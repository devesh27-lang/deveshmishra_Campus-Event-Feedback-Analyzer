## 2. Load your dataset

```python
df = pd.read_csv("Student_13_dataset.csv")
display(df.head())
```

## 3. Explore the data

```python
# TODO: Display the shape and column names
# TODO: Inspect data types and missing values
# TODO: Count duplicate rows
# Display the shape and column names
print('Shape:', df.shape)
print('Column names:', df.columns)

# Inspect data types
df.info()

# Check missing values in each column
print('Missing values:\n', df.isnull().sum())

# Count duplicate rows
print('Duplicate rows:', df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum
# Select the rating columns and convert them to a NumPy array
rating_columns = ['Content_Rating', 'Venue_Rating', 'Organization_Rating']
ratings = df[rating_columns].to_numpy()

# NumPy calculations
print('Overall average rating:', np.mean(ratings))
print('Highest rating:', np.max(ratings))
print('Lowest rating:', np.min(ratings))
print('Standard deviation:', np.std(ratings))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
# Add the total rating for each response
df['Total_Rating'] = df[rating_columns].sum(axis=1)

# Add the average rating for each response
df['Average_Rating'] = df[rating_columns].mean(axis=1).round(2)

# Filter responses with average rating 4 or above
high_rated = df[df['Average_Rating'] >= 4]
print('Responses with average rating >= 4:', len(high_rated))
display(high_rated)

# Sort all responses by average rating, highest first
sorted_responses = df.sort_values(by='Average_Rating', ascending=False)
display(sorted_responses.head())

# Display the lowest-rated responses
display(df.sort_values(by='Average_Rating').head())
```

## 6. Your individual project challenge
**Challenge:** Calculate average ratings; compare events; identify lowest-rated feedback area.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# Average score for each feedback area
area_averages = df[rating_columns].mean().round(2)
print('Average rating by feedback area:\n', area_averages)

# Compare events by average rating in each area
event_averages = df.groupby('Event')[rating_columns].mean().round(2)
event_averages['Overall_Event_Average'] = event_averages.mean(axis=1).round(2)
event_averages = event_averages.sort_values(by='Overall_Event_Average', ascending=False)
print('Event comparison (highest average first):')
display(event_averages)

# Identify the lowest-rated area (there may be a tie)
lowest_average = area_averages.min()
lowest_areas = area_averages[area_averages == lowest_average].index.tolist()
print('Lowest-rated feedback area(s):', lowest_areas)
print('Lowest average rating:', lowest_average)

# Identify the highest and lowest response records
print('Highest-rated response:')
display(df.sort_values(by='Average_Rating', ascending=False).head(1))
print('Lowest-rated response:')
display(df.sort_values(by='Average_Rating').head(1))
```
